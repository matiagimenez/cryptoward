# Cryptocurrency price scrapper

```python
import requests
import bs4

cryptocurrencies = ["bitcoin", "ethereum"]
base_url = "https://coinmarketcap.com/currencies"
for currency in cryptocurrencies:
    url = f"{base_url}/{currency}"
    response = requests.get(url)
    soup = bs4.BeautifulSoup(response.text, "html.parser")
    price_element = soup.find("span", attrs={"data-test": "text-cdp-price-display"})
    if price_element:
        price = price_element.text
        print(f"The current price of {currency.capitalize()} is: {price}")
    else:
        print(f"Price element not found for {currency}")
```
